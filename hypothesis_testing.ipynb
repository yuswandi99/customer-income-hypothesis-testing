# Uji Hipotesis

```python
from scipy.stats import kruskal
```

## H0 dan H1 (Pendidikan)
H0 : Tidak ada perbedaan income secara signifikan berdasarkan group pendidikan

H1 : Terdapat perbedaan income secara signifikan berdasarkan group pendidikan

```python
#memisahkan data income berdasarkan pendidikan
group_basic = df[df['Education']=='Basic']['Income']
group_2n = df[df['Education']=='2n Cycle']['Income']
group_grad = df[df['Education']=='Graduation']['Income']
group_master = df[df['Education']=='Master']['Income']
group_phd = df[df['Education']=='PhD']['Income']

#melakukan uji kruskal-wallis test
h_statistic, p_value_pendidikan = kruskal(
    group_basic,
    group_2n,
    group_grad,
    group_master,
    group_phd
)

print(f"H-Statistic : {h_statistic:.3f}")
print(f"P-Value     : {p_value_pendidikan:.4f}")
```

```python
alpha = 0.05

if p_value_pendidikan < alpha:
    print("Menerima H1")
else:
    print("Menerima H0")
```

## Interpretasi
- Menggunakan uji kruskal wallis test karena data tidak terdistribusi normal dan group independen lebih dari 3
- Berdasarkan hasil p-value maka H1 diterima karena nilai p-value < 0.05, Maka terdapat perbedaan income secara signifikan berdasarkan group pendidikan.
- Walaupun terdapat perbedaan income secara signifikan berdasarkan group pendidikan akan tetapi kita belum tau dimana titik perbedaannya maka perlu adanya post hoc test

```python
import scikit_posthocs as sp

df_education = df[['Education','Income']]
dunn = sp.posthoc_dunn(
    df_education,
    val_col="Income",
    group_col="Education",
    p_adjust="bonferroni"
)

print(dunn.map(lambda x: f"{x:.5f}"))
```

## Interpretasi

- Tingkat pendidikan Basic merupakan tingkat pendidikan yang memiliki perbedaan income secara signifikan terhadap semua tingkat pendidikan (p_value = 0.00). Berdasarkan rata-rata income diketahui bahwa basic memiliki rata-rata yang paling rendah.   
- Tingkat pendidikan Graduation dan Master tidak memiliki perbedaan income secara signifikan (p_value = 1.000).  
- Tingkat pendidikan Master dan PhD juga tidak memiliki perbedaan income secara signifikan (p_value = 0.41842).
- Sedangkan untuk tingkat pendidikan Graduation dan PhD memiliki perbedaan income yang signifikan (p_value = 0.02254). Karena diketahui bahwa rata-rata income PhD > Graduation, maka dapat disimpulkan bahwa kelompok PhD memiliki rata-rata diatas Graduation.
- Tingkat pendidikan 2n Cycle tidak memiliki perbedaan income secara signifikan dari Graduation dan Master (p > 0.05, meski mendekati batas), tapi terdapat perbedaan income secara signifikan terhadap Basic dan PhD.
- Maka berdasarkan kelompok pendidikan dapat dibagi menjadi tiga tingkat yaitu Grup rendah: Basic, Grup menengah: 2n Cycle, Grup tinggi: Graduation, Master, PhD



## H0 dan H1 (Marital Status)
H0 : Tidak ada perbedaan income berdasarkan marital status

H1 : Terdapat perbedaan income berdasarkan marital status

```python
#memisahkan data income berdasarkan marital status
group_single = df[df['Marital_Status']=='Single']['Income']
group_married = df[df['Marital_Status']=='Married']['Income']
group_divorced = df[df['Marital_Status']=='Divorced']['Income']
group_widow = df[df['Marital_Status']=='Widow']['Income']
group_together = df[df['Marital_Status']=='Together']['Income']

#melakukan uji kruskal-wallis test
h_statistic, p_value_marital = kruskal(
    group_single,
    group_married,
    group_divorced,
    group_widow,
    group_together
)

print(f"H-Statistic : {h_statistic:.3f}")
print(f"P-Value     : {p_value_marital:.4f}")
```

```python
alpha = 0.05

if p_value_marital < alpha:
    print("Menerima H1")
else:
    print("Menerima H0")
```

## Interpretasi
- Menggunakan uji kruskal wallis test karena data tidak terdistribusi normal dan group independen lebih dari 3 group
- Berdasarkan hasil p_value, nilai p_value > 0.05 ini berarti bahwa tidak ada perbedaan income berdasarkan marital status

## Uji Hipotesis
H0 : Tidak ada hubungan antara income dengan total spent

H1 : Terdapat hubungan antara income dengan total spent

```python
from scipy.stats import spearmanr

coef, p_val_numerik = spearmanr(df['Income'], df['total_spent'])


print(f"Koefisien Korelasi (Rho): {coef:.4f}")
print(f"P-Value                 : {p_val_numerik}")
```

```python
alpha = 0.05

if p_val_numerik < alpha:
    print("Menerima H1")
else:
    print("Menerima H0")
```

## Interpretasi
- Menggunakan uji Spearman Rank Correlation karena ingin menunjukkan apakah ada hubungan antara dua variabel, terdapat outlier dalam kedua variabel tersebut, dan kedua variabel tersebut tidak terdistribusi dengan normal.
- Berdasarkan hasil p_value, Maka H1 diterima karena p_value < 0.05. Dapat disimpulkan bahwa terdapat hubungan antara income dengan total spent
- Berdasarkan hasil coefisien, nilai coefisien bernilai 0.8473 ini menunjukkan bahwa korelasi antara hubungan tersebut adalah positif dan kuat.
- Kesimpulannya bahwa semakin tinggi income seseorang maka semakin banyak pula uang yang dihabiskan untuk berbelanja.

# Interpretasi dan Kesimpulan

- Karena Income berpengaruh positif dan kuat terhadap Total Spent, serta terdapat perbedaan signifikan Income antar tingkat pendidikan (Basic rendah, 2n Cycle menengah, dan Graduation/Master/PhD tinggi), maka dapat disimpulkan bahwa tingkat pendidikan secara tidak langsung berperan sebagai faktor pembeda dalam potensi pengeluaran pelanggan.
